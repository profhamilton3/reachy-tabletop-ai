```python
def show_position(name):
    pose = reachy.r_arm.forward_kinematics()

    x = pose[0, 3]
    y = pose[1, 3]
    z = pose[2, 3]

    print(
        f'{name}: '
        f'({x:.4f}, {y:.4f}, {z:.4f})'
    )

    return pose
```
